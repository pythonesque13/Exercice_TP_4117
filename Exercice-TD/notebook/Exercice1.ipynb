```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

points={
    'A1':(2,10),'A2':(2,5),'A3':(8,4),
    'B1':(5,8),'B2':(7,5),'B3':(6,4),
    'C1':(1,2),'C2':(4,9)
}

X=np.array(list(points.values()))
labels=list(points.keys())

print(X)
print(labels)
```

```python
kmeans=KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

cluster=kmeans.labels_
centroids=kmeans.cluster_centers_
```

```python
print("-- Resultats du CLustering --")
for i in range (3):
    print(f"Cluter {i}")
    cluster_points=[labels[j] for j, cluster_id in enumerate(cluster) if cluster_id == i]
    print(f" Points: {cluster_points}")
    print(f" Centroides: ({centroids[i][0]:.2f}, {centroids[i][1]:.2f})")
```

```python
plt.figure(figsize=(10, 8))
scatter=plt.scatter(X[:, 0], X[:, 1], c=cluster, cmap='viridis', s=100, alpha=0.7, edgecolors='k')

for i, label in enumerate(labels):
    plt.annotate(label, (X[i, 0] + 0.1, X[i, 1] + 0.1))
    plt.annotate(label, (X[i, 0] + 0.1, X[i, 1] + 0.1))

plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Centroides')


plt.title('Classification des points pour k=3')
plt.xlabel('En X')
plt.ylabel('En Y')
plt.legend()
plt.grid(True)
plt.show()
```
